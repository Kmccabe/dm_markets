```python
import matplotlib.pyplot as plt
```

```python
x1 = [600, 200]
x2 = [700, 300]
x3 = [800, 400]

y1 = [200, 600]
y2 = [300, 700]
y3 = [400, 800]


plt.plot(x1, label = 'Buyer 1')
plt.plot(x2, label = 'Buyer 2')
plt.plot(x3, label = 'Buyer 3')

plt.plot(y1, label = 'Seller 1')
plt.plot(y2, label = 'Seller 2')
plt.plot(y3, label = 'Seller 3')

plt.ylabel('Valuation/Cost')
plt.xlabel('Unit Space')

plt.legend(loc=(1, 0))
plt.show()
```

```python
# Simulate draws
import numpy as np
x1s = list(np.random.uniform(x1[0], x1[1], 80))
x2s = list(np.random.uniform(x2[0], x2[1], 80))
x3s = list(np.random.uniform(x3[0], x3[1], 80))
plt.hist(x1s+x2s+x3s)

s1s = list(np.random.uniform(y1[0], y1[1], 80))
s2s = list(np.random.uniform(y2[0], y2[1], 80))
s3s = list(np.random.uniform(y3[0], y3[1], 80))
plt.hist(s1s+s2s+s3s)

```
